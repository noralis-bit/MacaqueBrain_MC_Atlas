```r
library(reticulate)

use_virtualenv("/home/chaiqw/project/macaque_motor_new/.venv")
library(Seurat)
library(SeuratDisk)
library(anndata)
library(qs)
library(dplyr)
library(Matrix)
```

```r
path = '/data/data0-1/transfer/motor/stereo/macaque-20251017-allmotor/processed/'
save_path = '/mnt/90-connectome/Personal/CQW/motor_snRNA_new/allmotor_cellbins/'

# 获取路径下所有h5ad文件
h5ad_files <- list.files(path, pattern = "\\.h5ad$", full.names = TRUE)

# 读取每个h5ad文件并构建Seurat对象
seurat_objects <- list()

for (i in seq_along(h5ad_files)) {
    file_path <- h5ad_files[i]
    file_name <- basename(file_path)
    
    # 检查目标文件是否已存在
    output_file <- paste0(save_path, file_name)
    if (file.exists(output_file)) {
        cat("Skipped (already exists):", file_name, "\n")
        next
    }
    
    counts <- adata$X
    md <- adata$obs
    # 转换为Seurat对象
    seurat_obj <- CreateSeuratObject(
        counts = t(counts),
        meta.data = md
    )
    
    # 过滤细胞和基因
    seurat_obj <- subset(seurat_obj, 
                        subset = nFeature_RNA > 100
    )
    
    # 计算线粒体基因比例（仅对存在的基因计算）
    MTGenes <- c("ND6","COX1","ND5","ND4","ND2","ND4L","ATP8","CYTB","COX2","ND3","ATP6","ND1","COX3")
    mt_present <- intersect(MTGenes, rownames(seurat_obj))
    if (length(mt_present) > 0) {
      seurat_obj[["percent.mt"]] <- (colSums(x = GetAssayData(object = seurat_obj, assay = "RNA", layer = "counts")[mt_present, , drop = FALSE]) /
        seurat_obj[[paste0("nCount_","RNA")]] * 100)
    } else {
      seurat_obj[["percent.mt"]] <- 0
    }

    # 标准化和寻找高变基因
    seurat_obj <- NormalizeData(seurat_obj, normalization.method = "LogNormalize", scale.factor = 10000)

  
    ad_rna <- AnnData(
    X = t(seurat_obj@assays$RNA$data),
    obs = seurat_obj@meta.data
    )
    write_h5ad(ad_rna, output_file)

  # 存储到列表中
#   seurat_obj %>% saveRDS(file = paste0(save_path, '/', file_name, '.rds'))
  
  cat("Processed:", file_name, "\n")
}
```
